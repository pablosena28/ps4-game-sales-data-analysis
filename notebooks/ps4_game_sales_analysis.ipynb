# PS4 Game Sales — Análise Exploratória com Python

**Pergunta central:** como as vendas de jogos de PS4 se distribuem por título, gênero, publisher, região e período?


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/raw/ps4_games_sales.csv')
df.head()
```

## 1. Diagnóstico da base

```python
print('Dimensões:', df.shape)
print(df.isna().sum())
print('Duplicatas:', df.duplicated().sum())
```

## 2. Top 10 jogos

```python
df.nlargest(10, 'Global')[['Game','Year','Genre','Publisher','Global']]
```

## 3. Gêneros

```python
df.groupby('Genre')['Global'].agg(['count','sum','mean']).sort_values('sum', ascending=False).head(10)
```

## 4. Regiões

```python
regions = ['North America','Europe','Japan','Rest of World']
regional = df[regions].sum().sort_values(ascending=False)
regional
(regional / regional.sum() * 100).round(2)
```

## 5. Publishers

```python
df.dropna(subset=['Publisher']).groupby('Publisher')['Global'].agg(['count','sum','mean']).sort_values('sum', ascending=False).head(10)
```

## 6. Evolução anual — 2013 a 2018

```python
temporal = df[df['Year'].between(2013, 2018)]
temporal.groupby('Year')['Global'].agg(['count','sum','mean']).round(2)
```

## Síntese

- Grand Theft Auto V lidera as vendas globais registradas.
- Action lidera em volume total.
- Europa é a maior região da base em vendas somadas.
- Activision lidera entre publishers mantendo os rótulos originais.
- A análise temporal principal utiliza 2013–2018.
